```python
import numpy as np
import pandas as pd
import os
import pickle
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
input_folder  = "/home/mojtaba.kananisarche/Zh/UTA/feature_vectors/mobilenetv2"
pca_transformers_folder = f"{input_folder}/pca_transformers"

output_folder = f"{input_folder}/pca_transformed"
os.makedirs(output_folder, exist_ok=True)

for i in range(5):
    os.makedirs(os.path.join(output_folder, f'test_fold_{i+1}'), exist_ok=True)
    os.makedirs(os.path.join(output_folder, f'test_fold_{i+1}', 'train'), exist_ok=True)
    os.makedirs(os.path.join(output_folder, f'test_fold_{i+1}', 'test_validation'), exist_ok=True)

csv_files = [f for f in os.listdir(input_folder) if f.lower().endswith(('.csv'))]
print(len(csv_files))
```

```python
csv_files_folds = [[], [], [], [], []]
for csv_file in csv_files:
    csv_files_folds[int(csv_file[0])-1].append(csv_file)
csv_files_folds
```

```python
for i in range(len(csv_files_folds)):
    print(f"====={i+1}=====")
    
    with open(os.path.join(pca_transformers_folder, f"pca_{i+1}.pkl"), 'rb') as f:
        pca_model: PCA = pickle.load(f) 
    
    n_components = pca_model.n_components_
    column_names = [f'pc_{i+1}' for i in range(n_components)]
    
    print(f"# Components : {n_components}")
    
    print("train")
    for j in range(len(csv_files_folds)):
        if i == j : continue
        print(f"-----{j+1}-----")
        
        for csv_file in csv_files_folds[j]:
            output_file_path = os.path.join(output_folder, 
                                            f"test_fold_{i+1}",
                                            "train",
                                            f"{os.path.splitext(csv_file)[0]}_pca.csv")
            
            if os.path.exists(output_file_path):
                print(f"Skipping : file already exists {csv_file}")
                continue
                
            df = pd.read_csv(os.path.join(input_folder, csv_file), skiprows=0)
            transformed_df = pd.DataFrame(pca_model.transform(df.values), columns=column_names)
            transformed_df.to_csv(output_file_path, index_label='frame')
            
            print(f'Saved PCA transformed csv: {output_file_path}')
    
    print("test_validation")
    for csv_file in csv_files_folds[i]:
        output_file_path = os.path.join(output_folder, 
                                        f"test_fold_{i+1}",
                                        "test_validation",
                                        f"{os.path.splitext(csv_file)[0]}_pca.csv")

        if os.path.exists(output_file_path):
            print(f"Skipping : file already exists {csv_file}")
            continue

        df = pd.read_csv(os.path.join(input_folder, csv_file), skiprows=0)
        transformed_df = pd.DataFrame(pca_model.transform(df.values), columns=column_names)
        transformed_df.to_csv(output_file_path, index_label='frame')

        print(f'Saved PCA transformed csv: {output_file_path}')
```
