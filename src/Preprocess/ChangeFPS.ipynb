```python
import os
from concurrent.futures import ThreadPoolExecutor
import subprocess

def change_fps(input_path, output_path, target_fps=30):
    """
    Change the FPS of a single video using FFmpeg.
    """
    command = [
        "ffmpeg",
        "-i", input_path,
        "-r", str(target_fps),
        output_path
    ]
    try:
        subprocess.run(command, check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        print(f"Successfully processed: {input_path}")
    except subprocess.CalledProcessError as e:
        print(f"Error processing {input_path}: {e}")

def process_videos(input_folder, output_folder, target_fps=30):
    """
    Process all videos in a folder to adjust FPS, excluding already processed ones.
    """
    # Ensure the output folder exists
    os.makedirs(output_folder, exist_ok=True)

    # Gather all video files from the input folder
    video_files = [f for f in os.listdir(input_folder) if f.lower().endswith(('.mp4', '.mov', '.avi', '.MOV'))]

    # Define the task for each video, excluding already processed videos
    tasks = []
    for video_file in video_files:
        input_path = os.path.join(input_folder, video_file)
        output_file_name = os.path.splitext(video_file)[0] + "_30fps" + os.path.splitext(video_file)[1]
        output_path = os.path.join(output_folder, output_file_name)

        # Skip processing if the output file already exists
        if not os.path.exists(output_path):
            tasks.append((input_path, output_path))
        else:
            print(f"Skipping already processed file: {output_file_name}")
    
    for task in tasks:
        change_fps(task[0], task[1], target_fps=30)
        
#     # Process videos in parallel
#     with ThreadPoolExecutor() as executor:
#         futures = [executor.submit(change_fps, task[0], task[1], target_fps) for task in tasks]
#         for future in futures:
#             future.result()
```

```python
# Example usage
input_folder = "/work/long_lab/Mojtaba/data/DROZY/DROZY/videos_i8/"
output_folder = "/work/long_lab/Mojtaba/data/DROZY/DROZY/videos_i8/30fps"
process_videos(input_folder, output_folder, target_fps=30)
```
