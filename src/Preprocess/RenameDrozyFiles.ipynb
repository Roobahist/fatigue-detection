```python
label_txt_path = '/work/long_lab/Mojtaba/data/DROZY/DROZY/KSS.txt'
```

```python
import pandas as pd

data = []

with open(label_txt_path, "r") as file:
    for subject_id, line in enumerate(file):
        labels = line.strip().split()
        for video_id, label in enumerate(labels):
            data.append([subject_id+1, video_id+1, int(label)])

# Create DataFrame
df = pd.DataFrame(data, columns=["subject_id", "video_id", "label"])
```

```python
import os
import re

def rename_files(folder_path):
    for filename in os.listdir(folder_path):
        match = re.match(r"(\d+)-(\d+)_30fps_blendshape\.csv", filename)
        if match:
            subject_id = int(match.group(1))
            video_id = int(match.group(2))
            label = df[(df['subject_id'] == subject_id) & (df['video_id'] == video_id)]['label'].values[0]
            new_name = f"0_{subject_id + 60}_{label}_30fps_blendshape.csv"
            old_path = os.path.join(folder_path, filename)
            new_path = os.path.join(folder_path, new_name)
            os.rename(old_path, new_path)
            print(f"Renamed: {filename} -> {new_name}")

# Example usage
folder_path = "/work/long_lab/Mojtaba/data/DROZY/DROZY/videos_30fps/blendshapes/"  # Change this to the actual folder path
rename_files(folder_path)
```
