```python
import os
import pandas as pd

def consolidate_csv_files(root_directory):
    """
    Consolidates CSV files from multiple directories into a single DataFrame.

    Parameters:
        root_directory (str): The root directory containing F1, F2, F3, F4, F5 folders.

    Returns:
        pd.DataFrame: A consolidated DataFrame with an additional 'file_name' column.
    """
    all_data = []  # List to store data from all CSV files

    # Iterate through F1 to F5 folders
    for folder_name in ['F1', 'F2', 'F3', 'F4', 'F5']:
        summary_folder = os.path.join(root_directory, folder_name, 'Summary')
        
        # Check if the Summary folder exists
        if not os.path.exists(summary_folder):
            print(f"Summary folder not found in {folder_name}. Skipping...")
            continue
        
        # Iterate through CSV files in the Summary folder
        for csv_file in os.listdir(summary_folder):
            if csv_file.endswith('.csv'):
                csv_path = os.path.join(summary_folder, csv_file)
                
                # Load the CSV file into a DataFrame
                data = pd.read_csv(csv_path)
                
                # Add a column with the file name
                data['file_name'] = csv_file
                
                # Append to the list
                all_data.append(data)
    
    # Combine all DataFrames into a single DataFrame
    consolidated_data = pd.concat(all_data, ignore_index=True)
    
    return consolidated_data
```

```python
# Set the root directory
root_dir = 'Data/face_landmarks/'

# Consolidate the CSV files
consolidated_df = consolidate_csv_files(root_dir)

# Save the consolidated DataFrame to a CSV file (optional)
consolidated_df.to_csv('consolidated_dataset.csv', index=False)

# Print the first few rows
print(consolidated_df.head())
```
