```python
import os
import zipfile

# Folder containing the zip files
zip_folder = '.'

# Iterate over all files in the folder
for file_name in os.listdir(zip_folder):
    if file_name.endswith('.zip'):
        zip_file_path = os.path.join(zip_folder, file_name)
        extract_folder_name = os.path.splitext(file_name)[0]  # Remove the .zip extension
        extract_to_path = os.path.join(zip_folder, extract_folder_name)

        # Create the output folder if it doesn't exist
        os.makedirs(extract_to_path, exist_ok=True)

        # Extract the zip file
        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to_path)
        
        print(f"Extracted: {file_name} to {extract_to_path}")

print("All zip files extracted successfully!")
```

```python
import os
import shutil

# Paths
source_root = 'Extracted'  # Root folder containing F1P1, F1P2, etc.
destination_folder = 'Merged'  # Folder to copy videos to

# Create the destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

# Iterate through F1P1, F1P2, ..., F5P2
for folder_name in os.listdir(source_root):
    folder_path = os.path.join(source_root, folder_name)

    # Skip if it's not a folder
    if not os.path.isdir(folder_path):
        continue

    # Parse fold and part numbers (e.g., F1P1 -> Fold1_Part1)
    fold_number, part_number = folder_name[1], folder_name[3]
    inner_folder_name = f"Fold{fold_number}_part{part_number}"
    inner_folder_path = os.path.join(folder_path, inner_folder_name)

    # Skip if the expected folder doesn't exist
    if not os.path.isdir(inner_folder_path):
        print(f"Inner folder missing: {inner_folder_path}")
        continue

    # Iterate through subject folders (e.g., 21, 55)
    for subject_id in os.listdir(inner_folder_path):
        subject_folder_path = os.path.join(inner_folder_path, subject_id)

        # Skip if it's not a folder
        if not os.path.isdir(subject_folder_path):
            continue

        # Iterate through video files
        for video_name in os.listdir(subject_folder_path):
            video_path = os.path.join(subject_folder_path, video_name)

            # Check if the file is a video (mp4 or mov)
            if not video_name.endswith(('.mp4', '.mov', '.MOV')):
                continue

            # Build the new file name
            video_base, video_extension = os.path.splitext(video_name)
            new_video_name = f"{fold_number}_{subject_id}_{video_base}{video_extension.lower()}"
            new_video_path = os.path.join(destination_folder, new_video_name)

            # Copy the video to the destination folder
            shutil.copy(video_path, new_video_path)
            print(f"Copied: {video_path} -> {new_video_path}")

print("All videos have been successfully copied and renamed!")
```
