```python
import os
import pandas as pd

# Path to the main directory containing subfolders with CSV files
face_landmarks_dir = "face_landmarks"

# List all subfolders in the face_landmarks directory, skipping those starting with '.'
subfolders = [f.path for f in os.scandir(face_landmarks_dir) if f.is_dir() and not f.name.startswith(".")]

# Iterate through each subfolder
for subfolder in subfolders:
    print(f"Processing folder: {subfolder}")
    
    # Get a list of all CSV files in the subfolder
    csv_files = []
    for f in os.scandir(subfolder):
        if f.name.endswith(".csv"):
            csv_files.append(f.path)
    
    if not csv_files:
        print(f"No CSV files found in {subfolder}. Skipping.")
        continue
    
    print(f"Found {len(csv_files)} CSV files. Merging...")
    
    # Initialize an empty list to store dataframes
    dataframes = []
    for csv_file in csv_files:
        print(f"Reading file: {csv_file}")
        df = pd.read_csv(csv_file)
        dataframes.append(df)
    
    # Merge all the dataframes
    merged_df = pd.concat(dataframes, ignore_index=True)
    
    # Save the merged CSV
    merged_csv_path = os.path.join(subfolder, "merged_landmarks.csv")
    merged_df.to_csv(merged_csv_path, index=False)
    
    print(f"Merged {len(csv_files)} files. Merged CSV saved at: {merged_csv_path}")
```
