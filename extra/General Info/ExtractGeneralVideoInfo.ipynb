```python
import cv2
import pandas as pd
import os
import numpy as np

# Root directory containing all videos
merged_folder = 'Raw_30fps'

# Directory to save the extracted video information
video_info_output_path = 'video_info_30fps.csv'

# Initialize list to store video info
video_info_data = []

# Process all videos in the merged folder
for root, _, files in os.walk(merged_folder):
    for file_name in files:
        if file_name.endswith(('.mp4', '.mov', '.MOV')):  # Process only video files
            video_path = os.path.join(root, file_name)
            video_name, video_extension = os.path.splitext(file_name)
            
            print(f"Processing video: {video_name}")
            
            # Open the video file
            cap = cv2.VideoCapture(video_path)
            
            if not cap.isOpened():
                print(f"Error: Couldn't open video {video_path}")
                continue
            
            # Get video properties
            width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
            height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
            fps = cap.get(cv2.CAP_PROP_FPS)
            frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            duration = frame_count / fps  # Duration in seconds
            
            print(f"Width: {width}, Height: {height}, FPS: {fps}, Frame count: {frame_count}, Duration: {duration:.2f}s")
            
            # Initialize variables for mean brightness calculation
            total_brightness = 0
            sampled_frame_count = 0

            # Calculate the frame indices to sample (one every 60 seconds)
            frame_interval = int(fps * 60)  # 60 seconds * fps to get frame index increments
            total_samples = frame_count // frame_interval  # Number of samples (1 per minute)
            
            # Loop through the selected frame indices at 60-second intervals
            for i in range(total_samples):
                frame_index = i * frame_interval  # Calculate the frame index for the i-th minute
                
                # Set the video capture to the desired frame index
                cap.set(cv2.CAP_PROP_POS_FRAMES, frame_index)
                
                # Read the frame
                ret, frame = cap.read()
                if not ret:
                    break
                
                # Convert the frame to grayscale to calculate brightness
                gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
                total_brightness += np.sum(gray_frame)
                sampled_frame_count += 1

            # Calculate mean brightness
            mean_brightness = total_brightness / (sampled_frame_count * width * height) if sampled_frame_count > 0 else 0

            # Add extracted data to the video info list
            video_info_data.append({
                "video_name": video_name,
                "width": width,
                "height": height,
                "fps": fps,
                "frame_count": frame_count,
                "duration": duration,
                "mean_brightness": mean_brightness
            })
            
            cap.release()

# Save the collected video information to a CSV file
video_info_df = pd.DataFrame(video_info_data)
video_info_df.to_csv(video_info_output_path, index=False)
print(f"Video information saved to {video_info_output_path}")
```
