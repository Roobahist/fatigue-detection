```python
import pandas as pd
consolidated_df = pd.read_csv("consolidated_dataset.csv")
```

```python
consolidated_df['label'] = consolidated_df['file_name'].apply(lambda x: int(x.split('.csv')[0].split('_')[-1]))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

def apply_linear_regression_with_normalization(data):
    """
    Applies a linear regression model to the normalized dataset.
    
    Parameters:
        data (pd.DataFrame): Consolidated DataFrame with the target and predictors.
    
    Returns:
        dict: A dictionary containing model performance metrics and the fitted model.
    """
    # Separate predictors and target
    predictors = data.drop(columns=['label', 'file_name'])
    target = data['label']

    # Normalize predictors
    scaler = StandardScaler()
    normalized_predictors = scaler.fit_transform(predictors)

    # Split into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(
        normalized_predictors, target, test_size=0.2, random_state=42
    )

    # Initialize the linear regression model
    model = LinearRegression()

    # Train the model
    model.fit(X_train, y_train)

    # Make predictions on the test set
    y_pred = model.predict(X_test)

    # Calculate performance metrics
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    print(f"Model Performance:")
    print(f"Mean Squared Error (MSE): {mse:.4f}")
    print(f"R-squared (R²): {r2:.4f}")

    # Return the model, scaler, and metrics
    return {
        'model': model,
        'scaler': scaler,
        'mse': mse,
        'r2': r2,
        'coefficients': model.coef_,
        'intercept': model.intercept_
    }
```

```python
# Apply the regression model with normalization
results = apply_linear_regression_with_normalization(consolidated_df)

# Display results
print(f"Model Coefficients: {results['coefficients']}")
print(f"Model Intercept: {results['intercept']}")
```
